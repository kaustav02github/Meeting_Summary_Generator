```python
import whisper

model = whisper.load_model("base")
result = model.transcribe("/content/AlanCross-Porter-v2.mp3")
print(result["text"])
```

```python
!pip install -q groq
from groq import Groq
from google.colab import userdata
import os

# Load API key from Colab secrets
os.environ["GROQ_API_KEY"] = userdata.get("GROQ_API_KEY")
client = Groq(api_key=os.environ["GROQ_API_KEY"])

print("Groq API Key loaded:", bool(os.environ["GROQ_API_KEY"]))
```

```python
from groq import Groq
from google.colab import userdata
import os



print("Groq API key loaded:", bool(os.environ["GROQ_API_KEY"]))

# Assuming your transcript is already extracted
transcript = result["text"]

prompt = f"""
You are an expert meeting summarizer. Summarize the following meeting transcript clearly and concisely.

Provide:
1. 🗣 Main Discussion Points
2. ✅ Key Decisions
3. 📋 Action Items

Transcript:
{transcript}
"""

# ✅ Use the new Groq model
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",   # Changed to a currently supported model
    messages=[{"role": "user", "content": prompt}],
    temperature=0.3,
    max_tokens=500
)

summary = response.choices[0].message.content
print("---- MEETING SUMMARY ----\n")
print(summary)
```
